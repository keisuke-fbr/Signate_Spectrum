```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# 表示スタイル
sns.set_theme(style="whitegrid", context="talk")

# データ読み込み
df = pd.read_csv("data/train.csv", encoding="cp932")

# PCAの対象外カラム
exclude_cols = ["sample number", "species number", "樹種", "含水率"]
available_exclude_cols = [c for c in exclude_cols if c in df.columns]

# 指定カラムを除外してから、数値列のみをPCA対象にする
feature_df = df.drop(columns=available_exclude_cols, errors="ignore")
num_df = feature_df.select_dtypes(include=[np.number]).copy()

if num_df.shape[1] == 0:
    raise ValueError("PCA対象の数値列が見つかりません。")

# 欠損値補完（中央値）
num_df = num_df.fillna(num_df.median(numeric_only=True))

# 標準化
scaler = StandardScaler()
X_scaled = scaler.fit_transform(num_df)

# PCA（全成分）
pca = PCA()
X_pca = pca.fit_transform(X_scaled)

# 寄与率サマリ
explained = pca.explained_variance_ratio_
cum_explained = explained.cumsum()

summary = pd.DataFrame(
    {
        "component": np.arange(1, len(explained) + 1),
        "explained_variance_ratio": explained,
        "cumulative_explained_variance_ratio": cum_explained,
    }
)

# 閾値に必要な次元数
n90 = np.argmax(cum_explained >= 0.90) + 1
n95 = np.argmax(cum_explained >= 0.95) + 1
n99 = np.argmax(cum_explained >= 0.99) + 1

print("=== PCA Summary ===")
print(f"PCA除外カラム: {available_exclude_cols}")
print(f"PCA対象の元の次元数: {num_df.shape[1]}")
print(f"90%を説明する成分数: {n90}")
print(f"95%を説明する成分数: {n95}")
print(f"99%を説明する成分数: {n99}")
print("\n先頭10成分の寄与率:")
print(summary.head(10).to_string(index=False))

# 95%説明に必要な成分までのスコアを計算して表示
pca_95 = PCA(n_components=n95)
X_pca_95 = pca_95.fit_transform(X_scaled)
pca_cols = [f"PC{i+1}" for i in range(X_pca_95.shape[1])]
pca_scores_df = pd.DataFrame(X_pca_95, columns=pca_cols)

print("\n=== PCA Summary Table ===")
display(summary)

print("\n=== PCA Scores (first 10 rows) ===")
display(pca_scores_df.head(10))

# -----------------------------
# 深掘り考察: 特徴量の重要度とPCごとの寄与
# -----------------------------
loadings = pd.DataFrame(
    pca.components_.T,
    index=num_df.columns,
    columns=[f"PC{i+1}" for i in range(len(explained))],
)

# 全PCで寄与率重み付きの絶対負荷量を合計し、特徴量重要度とする
weighted_importance = (loadings.abs() * explained).sum(axis=1)
feature_importance_df = (
    weighted_importance.sort_values(ascending=False)
    .rename("weighted_importance")
    .reset_index()
    .rename(columns={"index": "feature"})
)

print("\n=== Feature Importance by PCA Loadings (Top 20) ===")
display(feature_importance_df.head(20))

# 上位PCについて、寄与の大きい特徴量を表示
n_show_pcs = min(5, len(explained))
top_k_each_pc = 10
pc_top_features = []

for i in range(n_show_pcs):
    pc_name = f"PC{i+1}"
    tmp = (
        loadings[pc_name]
        .abs()
        .sort_values(ascending=False)
        .head(top_k_each_pc)
        .to_frame(name="abs_loading")
    )
    tmp["loading"] = loadings.loc[tmp.index, pc_name].values
    tmp["component"] = pc_name
    tmp["explained_variance_ratio"] = explained[i]
    tmp = tmp.reset_index().rename(columns={"index": "feature"})
    pc_top_features.append(tmp)

pc_top_features_df = pd.concat(pc_top_features, ignore_index=True)

print(f"\n=== Top {top_k_each_pc} Features in Each of First {n_show_pcs} PCs ===")
display(pc_top_features_df)

pc_target_corr_df = None

# 含水率が存在する場合、PCA重要度上位特徴との相関、および各PCと含水率の相関を確認
if "含水率" in df.columns and pd.api.types.is_numeric_dtype(df["含水率"]):
    top_features = feature_importance_df["feature"].head(20).tolist()
    corr_df = pd.DataFrame(
        {
            "feature": top_features,
            "corr_with_含水率": [df[f].corr(df["含水率"]) for f in top_features],
        }
    )
    corr_df["abs_corr_with_含水率"] = corr_df["corr_with_含水率"].abs()
    corr_df = corr_df.sort_values("abs_corr_with_含水率", ascending=False)

    print("\n=== Correlation of Top PCA-Important Features with 含水率 ===")
    display(corr_df)

    # 各PC軸と含水率の相関（全軸）
    y = df["含水率"].copy()
    if y.isna().any():
        y = y.fillna(y.median())

    pc_target_corr_df = pd.DataFrame(
        {
            "component": [f"PC{i+1}" for i in range(X_pca.shape[1])],
            "corr_with_含水率": [pd.Series(X_pca[:, i]).corr(y) for i in range(X_pca.shape[1])],
        }
    )
    pc_target_corr_df["abs_corr_with_含水率"] = pc_target_corr_df["corr_with_含水率"].abs()

    print("\n=== Correlation Between Each PCA Axis and 含水率 (All PCs) ===")
    display(pc_target_corr_df)

    print("\n=== Top 10 PCA Axes by |Correlation with 含水率| ===")
    display(pc_target_corr_df.sort_values("abs_corr_with_含水率", ascending=False).head(10))

print("\n=== Feature Importance と Top 10 features の違い ===")
print("1) Feature Importance は、全PCを使って特徴量を総合評価（グローバルな重要度）したもの。")
print("2) Top 10 features は、各PCごとにその軸を作っている特徴量上位を見たもの（ローカルな重要度）。")
print("3) そのため、Feature Importance上位とPC別Top 10は一致する場合もありますが、目的が異なるため一致しないこともあります。")

# -----------------------------
# 可視化（rich）
# -----------------------------

# 1) Scree plot（寄与率 + 累積寄与率）
plot_n = min(30, len(explained))
fig, ax1 = plt.subplots(figsize=(14, 6))
ax2 = ax1.twinx()

x = np.arange(1, plot_n + 1)
ax1.bar(x, explained[:plot_n], alpha=0.7, color="#2A9D8F", label="Explained variance ratio")
ax2.plot(x, cum_explained[:plot_n], color="#E76F51", marker="o", label="Cumulative explained")

ax1.set_xlabel("Principal Component")
ax1.set_ylabel("Explained variance ratio")
ax2.set_ylabel("Cumulative explained variance ratio")
ax1.set_title("Scree Plot (First 30 PCs)")

for th, c in [(0.90, "#264653"), (0.95, "#1D3557"), (0.99, "#6D597A")]:
    ax2.axhline(th, linestyle="--", color=c, alpha=0.8)

ax1.grid(True, axis="y", alpha=0.25)
plt.tight_layout()
plt.show()

# 2) PC1-PC2散布図（含水率がある場合は色付け）
plot_df = pd.DataFrame({"PC1": X_pca[:, 0], "PC2": X_pca[:, 1]})

fig, ax = plt.subplots(figsize=(10, 8))
if "含水率" in df.columns and pd.api.types.is_numeric_dtype(df["含水率"]):
    scatter = ax.scatter(
        plot_df["PC1"],
        plot_df["PC2"],
        c=df["含水率"],
        cmap="viridis",
        s=26,
        alpha=0.85,
        edgecolors="none",
    )
    cbar = plt.colorbar(scatter, ax=ax)
    cbar.set_label("含水率")
else:
    ax.scatter(plot_df["PC1"], plot_df["PC2"], s=26, alpha=0.8, color="#457B9D", edgecolors="none")

ax.set_title("Score Plot: PC1 vs PC2")
ax.set_xlabel(f"PC1 ({explained[0] * 100:.2f}%)")
ax.set_ylabel(f"PC2 ({explained[1] * 100:.2f}%)")
ax.grid(alpha=0.2)
plt.tight_layout()
plt.show()

# 3) 特徴量重要度 Top 20
fi_top = feature_importance_df.head(20).copy()
fi_top = fi_top.sort_values("weighted_importance", ascending=True)

fig, ax = plt.subplots(figsize=(10, 9))
ax.barh(fi_top["feature"].astype(str), fi_top["weighted_importance"], color="#F4A261")
ax.set_title("Top 20 Feature Importance (PCA weighted loadings)")
ax.set_xlabel("weighted_importance")
ax.set_ylabel("feature")
plt.tight_layout()
plt.show()

# 4) 各PCのTop特徴量の負荷量ヒートマップ（PC1-PC5）
heat_rows = []
for i in range(n_show_pcs):
    pc = f"PC{i+1}"
    top_feats = (
        loadings[pc].abs().sort_values(ascending=False).head(top_k_each_pc).index.tolist()
    )
    for feat in top_feats:
        heat_rows.append({"feature": str(feat), "component": pc, "loading": loadings.loc[feat, pc]})

heat_df = pd.DataFrame(heat_rows)
heat_pivot = heat_df.pivot_table(index="feature", columns="component", values="loading", aggfunc="mean")

fig, ax = plt.subplots(figsize=(8, 10))
sns.heatmap(
    heat_pivot,
    cmap="coolwarm",
    center=0,
    linewidths=0.4,
    cbar_kws={"label": "loading"},
    ax=ax,
)
ax.set_title("Loadings Heatmap (Top features in PC1-PC5)")
ax.set_xlabel("component")
ax.set_ylabel("feature")
plt.tight_layout()
plt.show()

# 5) 含水率との相関が高いPCトップ10
if pc_target_corr_df is not None:
    corr_top10 = (
        pc_target_corr_df.sort_values("abs_corr_with_含水率", ascending=False)
        .head(10)
        .sort_values("corr_with_含水率", ascending=True)
    )

    fig, ax = plt.subplots(figsize=(9, 6))
    colors = ["#E63946" if v < 0 else "#2A9D8F" for v in corr_top10["corr_with_含水率"]]
    ax.barh(corr_top10["component"], corr_top10["corr_with_含水率"], color=colors)
    ax.axvline(0, color="black", linewidth=1)
    ax.set_title("Top 10 PCA Axes Correlated with 含水率")
    ax.set_xlabel("correlation with 含水率")
    ax.set_ylabel("component")
    plt.tight_layout()
    plt.show()
```

## スペクトル分析向けEDA（追加）

このセクションでは、以下を可視化します。
- 平均スペクトルとばらつき（全体）
- 含水率の四分位別スペクトル
- 樹種ごとの平均スペクトル（上位カテゴリ）
- Savitzky-Golay平滑化と1次微分スペクトル
- 含水率との波長ごとの相関プロファイル

```python
from scipy.signal import savgol_filter

# 前セルを未実行でも動くように最低限を再定義
if "df" not in globals():
    df = pd.read_csv("data/train.csv", encoding="cp932")

exclude_cols_local = ["sample number", "species number", "樹種", "含水率"]
feature_df_local = df.drop(columns=[c for c in exclude_cols_local if c in df.columns], errors="ignore")
num_df_local = feature_df_local.select_dtypes(include=[np.number]).copy()
num_df_local = num_df_local.fillna(num_df_local.median(numeric_only=True))

# 波長軸の推定（列名が数値化できるものを優先）
col_series = pd.Series(num_df_local.columns.astype(str), index=num_df_local.columns)
wavelength_map = pd.to_numeric(col_series, errors="coerce")
if wavelength_map.notna().sum() >= 50:
    spectral_cols = wavelength_map.dropna().sort_values().index.tolist()
    x_axis = wavelength_map.loc[spectral_cols].values
    x_label = "Wavelength"
else:
    spectral_cols = num_df_local.columns.tolist()
    x_axis = np.arange(len(spectral_cols))
    x_label = "Feature Index"

X_spec = num_df_local[spectral_cols]
mean_spec = X_spec.mean(axis=0).values
std_spec = X_spec.std(axis=0).values

# 1) 平均スペクトル + 標準偏差帯
fig, ax = plt.subplots(figsize=(14, 6))
ax.plot(x_axis, mean_spec, color="#1d3557", linewidth=2.2, label="Mean spectrum")
ax.fill_between(x_axis, mean_spec - std_spec, mean_spec + std_spec, color="#a8dadc", alpha=0.35, label="Mean ± 1SD")
ax.set_title("Global Mean Spectrum with Variability")
ax.set_xlabel(x_label)
ax.set_ylabel("Intensity")
ax.legend(loc="best")
ax.grid(alpha=0.2)
plt.tight_layout()
plt.show()

# 2) Moisture四分位ごとの平均スペクトル
if "含水率" in df.columns and pd.api.types.is_numeric_dtype(df["含水率"]):
    y_local = df["含水率"].copy()
    q_labels = ["Q1(low)", "Q2", "Q3", "Q4(high)"]
    q = pd.qcut(y_local.rank(method="first"), q=4, labels=q_labels)

    fig, ax = plt.subplots(figsize=(14, 6))
    palette = ["#264653", "#2a9d8f", "#e9c46a", "#e76f51"]
    for label, color in zip(q_labels, palette):
        idx = q == label
        if idx.sum() > 0:
            group_mean = X_spec.loc[idx].mean(axis=0).values
            ax.plot(x_axis, group_mean, linewidth=2, alpha=0.95, color=color, label=f"Moisture {label} (n={idx.sum()})")

    ax.set_title("Mean Spectrum by Moisture Quartile")
    ax.set_xlabel(x_label)
    ax.set_ylabel("Intensity")
    ax.legend(loc="best")
    ax.grid(alpha=0.2)
    plt.tight_layout()
    plt.show()

# 3) 樹種の上位カテゴリ別平均スペクトル
if "樹種" in df.columns:
    top_species = df["樹種"].astype(str).value_counts().head(5).index.tolist()
    fig, ax = plt.subplots(figsize=(14, 6))

    for sp in top_species:
        idx = df["樹種"].astype(str) == sp
        if idx.sum() > 0:
            sp_mean = X_spec.loc[idx].mean(axis=0).values
            ax.plot(x_axis, sp_mean, linewidth=1.8, alpha=0.9, label=f"{sp} (n={idx.sum()})")

    ax.set_title("Mean Spectrum by Top-5 Species")
    ax.set_xlabel(x_label)
    ax.set_ylabel("Intensity")
    ax.legend(loc="best")
    ax.grid(alpha=0.2)
    plt.tight_layout()
    plt.show()

# 4) Savitzky-Golay平滑化 + 1次微分（全体平均スペクトルで例示）
window = 11 if len(mean_spec) >= 11 else (len(mean_spec) // 2) * 2 + 1
window = max(window, 5)
if window >= len(mean_spec):
    window = len(mean_spec) - 1 if len(mean_spec) % 2 == 0 else len(mean_spec)
if window % 2 == 0:
    window -= 1

smoothed = savgol_filter(mean_spec, window_length=window, polyorder=2)
d1 = savgol_filter(mean_spec, window_length=window, polyorder=2, deriv=1)

fig, axes = plt.subplots(2, 1, figsize=(14, 9), sharex=True)
axes[0].plot(x_axis, mean_spec, color="#6c757d", alpha=0.7, label="Original mean")
axes[0].plot(x_axis, smoothed, color="#1d3557", linewidth=2, label="SG smoothed")
axes[0].set_title("Mean Spectrum: Original vs Savitzky-Golay Smoothed")
axes[0].set_ylabel("Intensity")
axes[0].legend(loc="best")
axes[0].grid(alpha=0.2)

axes[1].plot(x_axis, d1, color="#e63946", linewidth=1.8, label="First derivative")
axes[1].axhline(0, color="black", linewidth=1)
axes[1].set_title("First Derivative Spectrum (Savitzky-Golay)")
axes[1].set_xlabel(x_label)
axes[1].set_ylabel("d(Intensity)/dx")
axes[1].legend(loc="best")
axes[1].grid(alpha=0.2)

plt.tight_layout()
plt.show()
```

```python
# 5) 波長ごとのMoisture相関プロファイル（連続スペクトル上で可視化）
if "含水率" in df.columns and pd.api.types.is_numeric_dtype(df["含水率"]):
    y_local = df["含水率"].copy()
    if y_local.isna().any():
        y_local = y_local.fillna(y_local.median())

    corr_profile = X_spec.apply(lambda s: s.corr(y_local), axis=0)
    abs_corr_profile = corr_profile.abs().sort_values(ascending=False)

    # 相関プロファイル曲線
    fig, ax = plt.subplots(figsize=(14, 6))
    ax.plot(x_axis, corr_profile.values, color="#457b9d", linewidth=1.6)
    ax.axhline(0, color="black", linewidth=1)
    ax.set_title("Correlation Profile: Each Spectral Feature vs Moisture")
    ax.set_xlabel(x_label)
    ax.set_ylabel("Pearson correlation")
    ax.grid(alpha=0.2)
    plt.tight_layout()
    plt.show()

    # 絶対相関トップ20
    top_corr_df = pd.DataFrame(
        {
            "feature": abs_corr_profile.head(20).index.astype(str),
            "corr": corr_profile.loc[abs_corr_profile.head(20).index].values,
            "abs_corr": abs_corr_profile.head(20).values,
        }
    ).sort_values("abs_corr", ascending=True)

    fig, ax = plt.subplots(figsize=(10, 8))
    colors = ["#e63946" if v < 0 else "#2a9d8f" for v in top_corr_df["corr"]]
    ax.barh(top_corr_df["feature"], top_corr_df["corr"], color=colors)
    ax.axvline(0, color="black", linewidth=1)
    ax.set_title("Top 20 Spectral Features by |Correlation| with Moisture")
    ax.set_xlabel("Correlation with Moisture")
    ax.set_ylabel("Feature")
    plt.tight_layout()
    plt.show()

    display(top_corr_df.sort_values("abs_corr", ascending=False))

# 6) PCAスコア空間の外れ値候補（上位2軸）
if "X_pca" in globals():
    pc12 = pd.DataFrame({"PC1": X_pca[:, 0], "PC2": X_pca[:, 1]})
    z1 = (pc12["PC1"] - pc12["PC1"].mean()) / pc12["PC1"].std(ddof=0)
    z2 = (pc12["PC2"] - pc12["PC2"].mean()) / pc12["PC2"].std(ddof=0)
    dist = np.sqrt(z1**2 + z2**2)
    outlier_mask = dist >= np.quantile(dist, 0.99)

    fig, ax = plt.subplots(figsize=(10, 8))
    ax.scatter(pc12.loc[~outlier_mask, "PC1"], pc12.loc[~outlier_mask, "PC2"], s=18, alpha=0.45, color="#457b9d", label="normal")
    ax.scatter(pc12.loc[outlier_mask, "PC1"], pc12.loc[outlier_mask, "PC2"], s=35, alpha=0.9, color="#e63946", label="outlier candidate")
    ax.set_title("Potential Outliers in PCA Score Space (Top 1%)")
    ax.set_xlabel("PC1")
    ax.set_ylabel("PC2")
    ax.legend(loc="best")
    ax.grid(alpha=0.2)
    plt.tight_layout()
    plt.show()

    outlier_indices = pc12.index[outlier_mask].tolist()
    print(f"Outlier候補数 (上位1%): {len(outlier_indices)}")
    print("先頭20件のindex:", outlier_indices[:20])
```

## 追加実装: モデル比較（外れ値・微分・上位波長）

以下を5-fold CVで比較します。
- Baseline: Raw spectral features
- Outlier Removed: PCA外れ値候補を除外
- SG 1st Derivative: Savitzky-Golay 1次微分特徴
- Top-Corr Bands: 含水率との絶対相関上位150波長

```python
from sklearn.model_selection import KFold, GroupKFold
from sklearn.metrics import mean_squared_error
import lightgbm as lgb

# ターゲット準備
if "含水率" not in df.columns:
    raise ValueError("ターゲット列 '含水率' が見つかりません。")

y_full = df["含水率"].astype(float).copy()
if y_full.isna().any():
    y_full = y_full.fillna(y_full.median())

# スペクトル特徴（前セルのX_specを利用）
X_raw = X_spec.copy()

# SG 1次微分特徴
window_cv = 11 if X_raw.shape[1] >= 11 else (X_raw.shape[1] // 2) * 2 + 1
window_cv = max(window_cv, 5)
if window_cv >= X_raw.shape[1]:
    window_cv = X_raw.shape[1] - 1 if X_raw.shape[1] % 2 == 0 else X_raw.shape[1]
if window_cv % 2 == 0:
    window_cv -= 1

X_deriv = savgol_filter(X_raw.values, window_length=window_cv, polyorder=2, deriv=1, axis=1)
X_deriv = pd.DataFrame(X_deriv, columns=X_raw.columns, index=X_raw.index)

# 上位相関波長（絶対相関）
corr_for_select = X_raw.apply(lambda s: s.corr(y_full), axis=0).abs().sort_values(ascending=False)
top_k_bands = min(150, len(corr_for_select))
top_band_cols = corr_for_select.head(top_k_bands).index.tolist()
X_topband = X_raw[top_band_cols].copy()

# 外れ値除外データ
if "outlier_mask" in globals():
    keep_mask = ~outlier_mask
else:
    keep_mask = pd.Series([True] * len(X_raw), index=X_raw.index)

X_raw_in = X_raw.loc[keep_mask]
y_in = y_full.loc[keep_mask]

# GroupKFold（樹種がある場合）
use_group = "樹種" in df.columns
if use_group:
    groups_all = df["樹種"].astype(str)
    groups_in = groups_all.loc[keep_mask]


def cv_rmse_lgb(X, y, groups=None, seed=42):
    params = {
        "objective": "regression",
        "metric": "rmse",
        "verbosity": -1,
        "learning_rate": 0.03,
        "num_leaves": 31,
        "feature_fraction": 0.9,
        "bagging_fraction": 0.8,
        "bagging_freq": 1,
        "min_child_samples": 20,
        "n_estimators": 500,
        "random_state": seed,
    }

    if groups is None:
        splitter = KFold(n_splits=5, shuffle=True, random_state=seed)
        split_iter = splitter.split(X)
    else:
        splitter = GroupKFold(n_splits=5)
        split_iter = splitter.split(X, y, groups)

    fold_rmses = []
    for tr_idx, va_idx in split_iter:
        X_tr, X_va = X.iloc[tr_idx], X.iloc[va_idx]
        y_tr, y_va = y.iloc[tr_idx], y.iloc[va_idx]

        model = lgb.LGBMRegressor(**params)
        model.fit(X_tr, y_tr)
        pred = model.predict(X_va)
        rmse = np.sqrt(mean_squared_error(y_va, pred))
        fold_rmses.append(rmse)

    return np.mean(fold_rmses), np.std(fold_rmses), fold_rmses

results = []

mean_rmse, std_rmse, folds = cv_rmse_lgb(X_raw, y_full, groups_all if use_group else None)
results.append({
    "setting": "Baseline (Raw Spectra)",
    "n_samples": len(X_raw),
    "n_features": X_raw.shape[1],
    "cv_rmse_mean": mean_rmse,
    "cv_rmse_std": std_rmse,
})

mean_rmse, std_rmse, folds = cv_rmse_lgb(X_raw_in, y_in, groups_in if use_group else None)
results.append({
    "setting": "Outlier Removed (Top 1% in PC1-PC2)",
    "n_samples": len(X_raw_in),
    "n_features": X_raw_in.shape[1],
    "cv_rmse_mean": mean_rmse,
    "cv_rmse_std": std_rmse,
})

mean_rmse, std_rmse, folds = cv_rmse_lgb(X_deriv, y_full, groups_all if use_group else None)
results.append({
    "setting": "SG 1st Derivative",
    "n_samples": len(X_deriv),
    "n_features": X_deriv.shape[1],
    "cv_rmse_mean": mean_rmse,
    "cv_rmse_std": std_rmse,
})

mean_rmse, std_rmse, folds = cv_rmse_lgb(X_topband, y_full, groups_all if use_group else None)
results.append({
    "setting": f"Top-Corr Bands (Top {top_k_bands})",
    "n_samples": len(X_topband),
    "n_features": X_topband.shape[1],
    "cv_rmse_mean": mean_rmse,
    "cv_rmse_std": std_rmse,
})

result_df = pd.DataFrame(results).sort_values("cv_rmse_mean", ascending=True).reset_index(drop=True)

print("=== CV Comparison (Lower RMSE is better) ===")
display(result_df)

# 可視化
fig, ax = plt.subplots(figsize=(10, 5))
ax.barh(result_df["setting"], result_df["cv_rmse_mean"], xerr=result_df["cv_rmse_std"], color="#2a9d8f", alpha=0.85)
ax.invert_yaxis()
ax.set_xlabel("CV RMSE")
ax.set_ylabel("Setting")
ax.set_title("Model Comparison: Spectral Feature Strategies")
ax.grid(axis="x", alpha=0.2)
plt.tight_layout()
plt.show()

best_row = result_df.iloc[0]
print(f"Best setting: {best_row['setting']}")
print(f"RMSE: {best_row['cv_rmse_mean']:.5f} ± {best_row['cv_rmse_std']:.5f}")
```
