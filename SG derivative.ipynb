```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import savgol_filter

# 1. データの生成（2つの重なったピーク + ノイズ）
x = np.linspace(0, 10, 500)
true_signal = (np.exp(-(x - 4)**2 / 0.5) + 
               np.exp(-(x - 5)**2 / 0.8))
noisy_data = true_signal + np.random.normal(0, 0.05, len(x))

# 2. Savitzky-Golay法の適用
window = 51 # 窓幅
poly = 3    # 次数
smoothed = savgol_filter(noisy_data, window, poly, deriv=0)
der1     = savgol_filter(noisy_data, window, poly, deriv=1)
der2     = savgol_filter(noisy_data, window, poly, deriv=2)

# 3. 可視化
# 3つのグラフ領域を ax1, ax2, ax3 という名前で別々に受け取ります
fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(10, 12))

# 上段：平滑化（ノイズ除去）
ax1.plot(x, noisy_data, label='Noisy Data', alpha=0.3, color='gray')
ax1.plot(x, smoothed, label='SG Smoothed', color='blue', lw=2)
ax1.set_title("1. Smoothing (Removing Noise)")
ax1.legend()

# 中段：1階微分（傾き）
ax2.plot(x, der1, label='SG 1st Derivative', color='green')
ax2.axhline(0, color='black', lw=1, ls='--') # 0の線
ax2.set_title("2. 1st Derivative (Slope)")
ax2.legend()

# 下段：2階微分（ピークの分離！）
# 見やすくするためにマイナスを掛けて「上向きの針」にしています
ax3.plot(x, -der2, label='SG 2nd Derivative (Inverted)', color='red')
ax3.set_title("3. 2nd Derivative (Resolving Peaks!)")
ax3.legend()

plt.tight_layout()
plt.show()
```
